```python
from deep_translator import GoogleTranslator

# Initialize the translator
translator = GoogleTranslator(source='en', target='fa')

# Function to translate subtitles
def translate_srt(file_path, output_file):
    with open(file_path, 'r', encoding='utf-8') as file:
        lines = file.readlines()

    translated_lines = []
    for line in lines:
        # Translate only the subtitle text (skip timing lines and indices)
        if line.strip().isdigit() or "-->" in line:
            translated_lines.append(line)
        elif line.strip() != "":
            try:
                # Translate the text
                translated_text = translator.translate(line.strip())
                translated_lines.append(translated_text + '\n')
            except Exception as e:
                print(f"Error translating line: {line.strip()} - {e}")
                translated_lines.append(line)  # If translation fails, keep the original line
        else:
            translated_lines.append(line)

    # Save translated subtitles to a new file
    with open(output_file, 'w', encoding='utf-8') as output:
        output.writelines(translated_lines)

# Translate 'zz.srt' to Persian and save as 'translated_zz.srt'
translate_srt('/content/zz.srt', 'translated_zz.srt')
```
